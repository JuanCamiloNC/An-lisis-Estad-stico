### Prueba t para diferencia de medias

```python
import pandas as pd
from scipy import stats
from statsmodels.stats.proportion import proportions_ztest
```

```python
df = pd.read_csv('Base_de_Datos_dynamic_pricing.csv')
```

```python
df.head()
```

```python
regular = df[df['Customer_Loyalty_Status'] == 'Regular']['Expected_Ride_Duration']
gold    = df[df['Customer_Loyalty_Status'] == 'Gold']['Expected_Ride_Duration']
```

```python
t_stat, p_valor = stats.ttest_ind(regular, gold, equal_var=False)
```

```python
print(f"Media Regular: {regular.mean():.2f}")
print(f"Media Gold:    {gold.mean():.2f}")
print(f"Estadístico t: {t_stat:.4f}")
print(f"p-valor:       {p_valor:.4f}")
```

### Prueba t para diferencia de proporciones

```python
df['rating_alto'] = (df['Average_Ratings'] >= 4.0).astype(int)
```

```python
economy = df[df['Vehicle_Type'] == 'Economy']['rating_alto']
premium = df[df['Vehicle_Type'] == 'Premium']['rating_alto']
```

```python
counts = [economy.sum(), premium.sum()]
nobs   = [len(economy), len(premium)]
```

```python
z_stat, p_valor = proportions_ztest(counts, nobs)
```

```python
print(f"Proporción Economy: {economy.mean():.4f}")
print(f"Proporción Premium: {premium.mean():.4f}")
print(f"Estadístico z: {z_stat:.4f}")
print(f"p-valor:      {p_valor:.4f}")
```

## Interpretación de las pruebas de hipótesis

En este análisis se realizaron dos comparaciones entre grupos utilizando un nivel de significancia de **α = 0.05**. El objetivo fue determinar si existen diferencias estadísticamente significativas en la duración esperada de los viajes según el nivel de fidelidad del cliente y en la proporción de calificaciones altas según el tipo de vehículo.

### 1. Comparación de la duración esperada según la fidelidad del cliente

Se aplicó una prueba t de Welch para comparar la media de `Expected_Ride_Duration` entre los clientes con estado de fidelidad **Regular** y **Gold**. Las hipótesis planteadas fueron:

- **H₀:** las medias de la duración esperada son iguales para los clientes Regular y Gold (`μRegular = μGold`).
- **H₁:** las medias de la duración esperada son diferentes (`μRegular ≠ μGold`).

La media de duración esperada para los clientes Regular fue de **100.57 minutos**, mientras que para los clientes Gold fue de **101.60 minutos**. La diferencia observada fue de aproximadamente **1.03 minutos**. El estadístico de prueba fue **t = -0.2624** y el p-valor fue **0.7931**.

Como el p-valor es mayor que 0.05, no se rechaza la hipótesis nula. Por tanto, con la información disponible no existe evidencia estadística suficiente para afirmar que la duración esperada promedio difiera entre los clientes Regular y Gold. Aunque las medias muestrales no son idénticas, la diferencia observada es pequeña en relación con la variabilidad de los datos y puede explicarse por fluctuaciones propias del muestreo.

### 2. Comparación de proporciones de calificaciones altas según el vehículo

Para esta comparación se creó la variable `rating_alto`, que toma el valor 1 cuando `Average_Ratings` es mayor o igual a 4.0 y el valor 0 en caso contrario. Posteriormente se comparó la proporción de calificaciones altas entre los vehículos **Economy** y **Premium** mediante una prueba de dos proporciones. Las hipótesis fueron:

- **H₀:** la proporción de calificaciones altas es igual en ambos tipos de vehículo (`pEconomy = pPremium`).
- **H₁:** las proporciones de calificaciones altas son diferentes (`pEconomy ≠ pPremium`).

La proporción de calificaciones altas fue de **0.6967**, equivalente al **69.67 %**, para los vehículos Economy. En los vehículos Premium fue de **0.6762**, equivalente al **67.62 %**. La diferencia entre ambas proporciones fue de aproximadamente **0.0205**, es decir, **2.05 puntos porcentuales**. El estadístico obtenido fue **z = 0.6946** y el p-valor fue **0.4873**.

Dado que el p-valor también es mayor que 0.05, no se rechaza la hipótesis nula. En consecuencia, no se encuentra evidencia estadística suficiente para afirmar que el tipo de vehículo esté asociado con una diferencia en la proporción de calificaciones altas. La mayor proporción observada en Economy es descriptiva, pero no es suficientemente grande para considerarse una diferencia significativa desde el punto de vista estadístico.
