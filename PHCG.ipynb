### Prueba t para diferencia de medias

```python
import pandas as pd
from scipy import stats
from statsmodels.stats.proportion import proportions_ztest
```

```python
df = pd.read_csv('Base_de_Datos_dynamic_pricing.csv')
```

```python
df.head()
```

```python
regular = df[df['Customer_Loyalty_Status'] == 'Regular']['Expected_Ride_Duration']
gold    = df[df['Customer_Loyalty_Status'] == 'Gold']['Expected_Ride_Duration']
```

```python
t_stat, p_valor = stats.ttest_ind(regular, gold, equal_var=False)
```

```python
print(f"Media Regular: {regular.mean():.2f}")
print(f"Media Gold:    {gold.mean():.2f}")
print(f"Estadístico t: {t_stat:.4f}")
print(f"p-valor:       {p_valor:.4f}")
```

### Prueba t para diferencia de proporciones

```python
df['rating_alto'] = (df['Average_Ratings'] >= 4.0).astype(int)
```

```python
economy = df[df['Vehicle_Type'] == 'Economy']['rating_alto']
premium = df[df['Vehicle_Type'] == 'Premium']['rating_alto']
```

```python
counts = [economy.sum(), premium.sum()]
nobs   = [len(economy), len(premium)]
```

```python
z_stat, p_valor = proportions_ztest(counts, nobs)
```

```python
print(f"Proporción Economy: {economy.mean():.4f}")
print(f"Proporción Premium: {premium.mean():.4f}")
print(f"Estadístico z: {z_stat:.4f}")
print(f"p-valor:      {p_valor:.4f}")
```
